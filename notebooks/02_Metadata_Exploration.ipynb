```python
metadata = sample_df.set_index(0).T
metadata.head()
```

```python
# 1. Identify column indices where the number of unique values is greater than 2
keep_columns = [
    i for i in range(metadata.shape[1]) 
    if metadata.iloc[:, i].nunique() > 2
]

# 2. Subset the dataframe by those positions using .iloc
metadata1 = metadata.iloc[:, keep_columns]

# View the result
metadata1
```

```python
import pandas as pd

# 1. Create a list for the brand new column names
new_headers = []
num_cols = metadata1.shape[1]

for i in range(num_cols):
    # Safely look at the first data row (index 0) using integer positioning
    first_val = str(metadata1.iloc[0, i])
    
    if ":" in first_val:
        # '"age: 32"' becomes 'age'
        clean_header = first_val.split(":")[0].replace('"', '').strip()
        new_headers.append(clean_header)
    else:
        # Fallback to the current column name string if no colon is present
        fallback = str(metadata1.columns[i]).replace('"', '').strip()
        new_headers.append(fallback if fallback else f"column_{i}")

# 2. Clean the data cells column-by-column using integer locations to avoid duplicate traps
for i in range(num_cols):
    # Grab the single column as a Series safely
    col_series = metadata1.iloc[:, i].astype(str)
    
    # Process the text: split on the colon and strip away quotes/spaces
    cleaned_series = col_series.str.split(':', n=1).str[-1]
    cleaned_series = cleaned_series.str.replace('"', '', regex=False).str.strip()
    
    # Assign the cleaned data back to its exact position
    metadata1.iloc[:, i] = cleaned_series

# 3. Assign the beautifully extracted headers directly to the dataframe
metadata1.columns = new_headers

# Note: Based on your latest screenshot, your data rows ALREADY start at index 0 
# (there is no extra "!Sample..." row at index 0 inside the data rows anymore).
# We just reset the index to keep it clean.
metadata1 = metadata1.reset_index(drop=True)

# Check the beautiful new result
print(metadata1.head())
```
