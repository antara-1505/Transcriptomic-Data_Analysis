```python
metadata2 = metadata1.drop(columns=[col for col in metadata1.columns if metadata1[col].nunique() <= 2])
metadata2
```

```python
import pandas as pd

# 1. Create a list for the brand new column names
new_headers = []

for col in metadata2.columns:
    # Look at the first data row (index 1) to find the type prefix
    # e.g., '"age: 32"'
    first_val = str(metadata2[col].iloc[1])
    
    if ":" in first_val:
        # Split by the colon, take the left side, and clean up quotes/spaces
        # '"age: 32"' becomes 'age'
        clean_header = first_val.split(":")[0].replace('"', '').strip()
        new_headers.append(clean_header)
    else:
        # If there's no colon (like an ID column), keep its original name or row 0 name
        fallback = str(metadata2[col].iloc[0]).replace('"', '').strip()
        new_headers.append(fallback if fallback else col)

# 2. Clean the data cells: Remove the "prefix: " part and outer quotes from all rows
for col in metadata2.columns:
    # Remove everything up to the first colon
    metadata2[col] = metadata2[col].astype(str).str.split(':', n=1).str[-1]
    # Strip any remaining double quotes and whitespaces
    metadata2[col] = metadata2[col].str.replace('"', '', regex=False).str.strip()

# 3. Assign the beautifully extracted headers directly to the dataframe
metadata2.columns = new_headers

# 4. Drop the original row 0 since it just contains the old "!Sample..." tags
metadata2 = metadata2.iloc[1:].reset_index(drop=True)

# Check the result
# metadata2 = pd.DataFrame(metadata2).T  # Ensure it's a DataFrame in case of any type issues
print(metadata2.head())
```
